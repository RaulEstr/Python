## 1) Por qué son necesarios

Sin aislamiento, todos los proyectos comparten paquetes globales y aparecen conflictos de versiones.

Problemas típicos:
- Proyecto A requiere Flask 2.x y Proyecto B Flask 3.x.
- Una actualización rompe prácticas antiguas.
- En otro ordenador no funciona igual porque no tiene las mismas versiones.

Un entorno virtual separa dependencias por proyecto y permite reproducir el entorno con precisión.

````python
import sys

print('Python ejecutable actual:', sys.executable)
print('prefix:', sys.prefix)
print('base_prefix:', sys.base_prefix)
print('Dentro de un entorno virtual:', sys.prefix != sys.base_prefix)
````

## 2) Tipos de herramientas

Resumen rápido:
- venv: estándar de Python, simple y muy adecuado para empezar.
- virtualenv: opción histórica con compatibilidad amplia.
- conda: entorno + paquetes Python y no Python.
- pipenv y poetry: gestión de dependencias de más alto nivel.
- uv: herramienta moderna, muy rápida, con gestión de entornos y paquetes.

## 3) Flujo clásico con venv (paso a paso)

### Paso 1. Crear carpeta de proyecto
```bash
mkdir mi_proyecto
cd mi_proyecto
```

### Paso 2. Crear entorno
```bash
python -m venv .venv
```

### Paso 3. Activar entorno
macOS/Linux:
```bash
source .venv/bin/activate
```
Windows PowerShell:
```powershell
.venv\Scripts\Activate.ps1
```

### Paso 4. Instalar paquetes
```bash
python -m pip install --upgrade pip
pip install flask pytest
```

### Paso 5. Guardar dependencias
```bash
pip freeze > requirements.txt
```

## 4) Flujo moderno con uv (paso a paso)

uv destaca por velocidad y simplicidad para gestión diaria.

### Paso 1. Instalar uv
macOS con Homebrew:
```bash
brew install uv
```
Instalador oficial multiplataforma:
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

### Paso 2. Crear entorno con uv
```bash
uv venv .venv
```
Con versión concreta de Python:
```bash
uv venv --python 3.12 .venv
```

### Paso 3. Activar entorno
macOS/Linux:
```bash
source .venv/bin/activate
```
Windows PowerShell:
```powershell
.venv\Scripts\Activate.ps1
```

### Paso 4. Instalar dependencias con uv
```bash
uv add flask pytest
```

### Paso 5. Exportar requirements para entornos tradicionales
```bash
uv pip freeze > requirements.txt
```

## 5) Cuándo usar venv y cuándo uv

Guía rápida:
- Usa venv si quieres partir del estándar oficial de Python.
- Usa uv si quieres un flujo moderno y más rápido para proyectos nuevos.
- En ambos casos, mantén aislamiento por proyecto y control de versiones.

````python
import sys

def estoy_en_entorno_virtual() -> bool:
    return sys.prefix != sys.base_prefix

print('Entorno activo:', estoy_en_entorno_virtual())
````

## 6) Errores frecuentes y solución

1. Se instalan paquetes fuera del entorno activo.
   Solución: verificar activación y usar python -m pip o uv add según flujo.
2. No activa en PowerShell.
   Solución: revisar ExecutionPolicy del equipo.
3. ModuleNotFoundError al lanzar tests.
   Solución: reinstalar dependencias y comprobar entorno activo.
4. Proyecto no reproducible en otro equipo.
   Solución: congelar dependencias en requirements.txt y reconstruir entorno.
