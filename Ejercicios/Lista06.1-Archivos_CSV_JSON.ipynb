# EJERCICIOS CSV

## Ejercicio 1 – Crear y leer un archivo CSV

Objetivo: Crear un archivo CSV con datos de alumnos con Python y mostrar su contenido con Python. (El CSV no se debe crear fuera del código).

## Ejercicio 2.1  – Filtrar registros de un CSV (con `DictReader`)

Objetivo: Mostrar solo los alumnos mayores de 20 años utilizando `DictReader`.

## Ejercicio 2.2  – Filtrar registros de un CSV (sin `DictReader`)

Objetivo: Mostrar solo los alumnos mayores de 20 años sin utilizar `DictReader`.

## Ejercicio 3 – Calcular un promedio desde CSV

Objetivo: Leer las edades del archivo y calcular la media.

## Ejercicio 4 – Mostrar CSV con pandas

Objetivo: Cargar el CSV en un DataFrame y mostrar información.

# EJERCICIOS JSON

## Ejercicio 1 – Crear y leer un archivo JSON

Objetivo: Guardar información de una clase y leerla.

## Ejercicio 2 – Agregar un nuevo registro al JSON

Objetivo: Cargar el archivo, añadir un nuevo alumno y volver a guardarlo.

## Ejercicio 3 – Calcular edad promedio desde JSON

Objetivo: Leer el archivo y calcular la edad media de los alumnos.

## Ejercicio 4 – Convertir datos JSON a CSV

Objetivo: Leer el archivo JSON y exportar su contenido a un CSV.
