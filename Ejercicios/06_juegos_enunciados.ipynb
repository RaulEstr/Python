### Juego 1: Adivina el número

El ordenador piensa un número al azar entre 1 y 100. El jugador intenta adivinarlo; tras cada intento el programa dice si el número secreto es **mayor** o **menor**. Al acertar, se muestra cuántos intentos ha necesitado y la lista con todos ellos.

**Practica**: `while`, `if/elif/else`, comparadores, listas y `random.randint`.

```python
# Escribe aquí tu solución
import random

def pedirNumero():
    numeroCorrecto = False
    while not numeroCorrecto:
        numero = (int)(input("Introduce un numero entre el 1 y el 100"))
        if numero < 1 or numero > 100:
            print("Tienes que introducir un numero entre 1 y 100")
        else:
            numeroCorrecto = True
    return numero
    

secreto = random.randint(1,100)
nums = []
acertado = False
while not acertado:
    numeroPedido = pedirNumero()
    nums.append(numeroPedido)
    if(secreto == numeroPedido):
        acertado = True
    elif(numeroPedido < secreto):
        print("El numero es mayor")
    else:
        print("El numero es menor")

print("Has acertado el numero {0} en {1} intentos".format(secreto, len(nums)))
print("Tus intentos: {0}".format(nums))
```

### Juego 2: Ahorcado

Elige una palabra al azar de una lista. El jugador va diciendo letras: si acierta se destapa la letra, si falla pierde una vida. Tiene **6 fallos** antes de perder. Muestra la palabra tapada (`_ _ _`) en cada turno.

**Practica**: `while`, sets (letras acertadas), `in`, recorrer cadenas y `random.choice`.

```python
# Escribe aquí tu solución
import random

PALABRAS = ["perro", "gato", "pajaro"]
LETRAS = ["a","b","c","d","e","f","g","h","i","j","k","l","m","n","ñ","o","p","q","r","s","t","u","v","w","x","y","z"]
MAX_INTENTOS = 6

palabra = random.choice(PALABRAS)

palabraAdivinada = ""
for i in range(len(palabra)):
    palabraAdivinada += "_ "

letrasIntroducidas = set()
cont = 0
acertado = False
while not acertado and cont < MAX_INTENTOS:
    print(palabraAdivinada)

    letra = input("Introduce una letra").lower()
    while letra not in LETRAS:
        letra = letra = input("Introduce una letra").lower()

    letrasIntroducidas.add(letra)

    if letra in palabra:
        palabraAdivinada = ""
        for i in range(len(palabra)):
            if palabra[i] in letrasIntroducidas:
                palabraAdivinada += palabra[i] + " "
            else:
                palabraAdivinada += "_ "        

    if "_" not in palabraAdivinada:
        acertado = True
        print(palabraAdivinada)
        print("Has acertado la palabra " + palabra)
          
    cont += 1

if cont == MAX_INTENTOS:
    print(palabraAdivinada)
    print("No has acertado, la palabra era " + palabra)
```

### Juego 3: La carrera de los dados

Tú y la máquina competís por llegar a la meta (casilla 20). En cada turno ambos tiráis un dado del 1 al 6 y avanzáis. Gana quien llegue primero a la meta.

**Practica**: `while`, comparadores, contadores y `random.randint`.

```python
# Escribe aquí tu solución
import random

casillaJugador = 1
casillaCPU = 1
llegado = False
while casillaJugador < 20 and casillaCPU < 20:
    input("Tira d6")

    tiradaJugador = random.randint(1,6)
    tiradaCPU = random.randint(1,6)

    casillaJugador += tiradaJugador
    casillaCPU += tiradaCPU

    print("Has sacado {0}, avanzas a la casilla {1}".format(tiradaJugador, casillaJugador))
    print("CPU ha sacado {0}, avanza a la casilla {1}".format(tiradaCPU, casillaCPU))
    print("")

if casillaJugador > casillaCPU:
    print("¡Has ganado!")
elif casillaCPU > casillaJugador:
    print("¡Has perdido!")
else:
    print("¡Empate!")
```

### Juego 4: Piedra, papel o tijera (una ronda)

El jugador elige piedra, papel o tijera y la máquina saca una al azar. Muestra quién ha ganado la ronda. Valida que la opción sea correcta.

**Practica**: funciones, condicionales con `and`, `in` y `random.choice`.

```python
# Escribe aquí tu solución
import random

def dameResultado(opcionJugador, opcionCPU):
    """
    Función que devuelve "Victoria" "Derrota" o "Empate" en funcion de las opciones que se le pasan como parametro
    Params:
        opcionJugador: str -> Opcion del jugador
        opcionCPU: str -> Opcion de la CPU

    Returns:
        "Victoria", "Derrota" o "False"

    """
    if opcionJugador == "Piedra":
        if opcionCPU == "Piedra":
            return "Empate"
        elif opcionCPU == "Papel":
            return "Derrota"
        else:
            return "Victoria"
    elif opcionJugador == "Papel":
        if opcionCPU == "Piedra":
            return "Victoria"
        elif opcionCPU == "Papel":
            return "Empate"
        else:
            return "Derrota"
    else:
        if opcionCPU == "Piedra":
            return "Derrota"
        elif opcionCPU == "Papel":
            return "Victoria"
        else:
            return "Empate"

NUM_RONDAS = 3
victoriasJugador = 0
victoriasCPU = 0
opciones = ["Piedra", "Papel", "Pijeras"]

while victoriasJugador < NUM_RONDAS and victoriasCPU < NUM_RONDAS:
    
    eleccionCPU = random.choice(opciones)
    eleccionJugador = input("Elige: Piedra, Papel o Tijeras").capitalize()
    while eleccionJugador not in opciones:
        print("No es una opcion valida")
        eleccionJugador = input("Elige: Piedra, Papel o Tijeras").capitalize() 
    
    resultado = dameResultado(eleccionJugador, eleccionCPU)
    if resultado == "Empate":
        print("{0} vs {1} --> EMPATE".format(eleccionJugador, eleccionCPU))
    elif resultado == "Victoria":
        print("{0} vs {1} --> VICTORIA".format(eleccionJugador, eleccionCPU))
        victoriasJugador += 1
    else:
        print("{0} vs {1} --> DERROTA".format(eleccionJugador, eleccionCPU))
        victoriasCPU += 1

if victoriasJugador == 3:
    print("¡GAS GANADO! c:")

if victoriasCPU == 3:
    print("¡HAS PERDIDO! :c")

```
