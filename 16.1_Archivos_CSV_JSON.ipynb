# Ficheros CSV y JSON en Python
En este cuaderno aprenderás cómo leer, escribir y procesar archivos **CSV** y **JSON** en Python, utilizando las bibliotecas estándar `csv`, `json` y la popular librería `pandas`.

## Introducción: ¿Qué son CSV y JSON?
- **CSV (Comma-Separated Values)**: formato de texto con filas separadas por saltos de línea y columnas por comas.
- **JSON (JavaScript Object Notation)**: formato estructurado usado para representar datos en forma de diccionarios y listas.

📍 **Cuándo usar cada uno:**
- CSV: ideal para datos tabulares (hojas de cálculo, bases de datos planas).
- JSON: útil para datos jerárquicos o con estructuras complejas (configuraciones, APIs, objetos).

## Lectura de archivos CSV con `csv.reader`
Permite leer archivos delimitados (por comas, punto y coma, etc.). Devuelve un iterador con cada fila como una lista.

```python
import csv

# Crear un archivo CSV de ejemplo
with open("alumnos.csv", "w", newline='', encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Nombre", "Edad", "Curso"])
    writer.writerow(["Ana", 20, "Python"])
    writer.writerow(["Luis", 22, "JavaScript"])
    writer.writerow(["María", 21, "Java"])

# Leer el archivo CSV
with open("alumnos.csv", "r", encoding="utf-8") as f:
    lector = csv.reader(f)
    for fila in lector:
        print(fila)
```

## Escritura de archivos CSV con `csv.writer`
Se utiliza para escribir filas en un archivo CSV con `writerow()` o varias con `writerows()`.

```python
alumnos = [
    ["Pedro", 23, "C++"],
    ["Lucía", 20, "Python"],
    ["Andrés", 22, "HTML"]
]

with open("nuevos_alumnos.csv", "w", newline='', encoding="utf-8") as f:
    escritor = csv.writer(f)
    escritor.writerow(["Nombre", "Edad", "Curso"])
    escritor.writerows(alumnos)

print("Archivo 'nuevos_alumnos.csv' creado correctamente.")
```

## Lectura y escritura de CSV con diccionarios (`DictReader` y `DictWriter`)
Estos métodos permiten acceder a los valores por nombre de columna, lo que hace el código más claro y legible.

```python
# Lectura con DictReader
with open("alumnos.csv", "r", encoding="utf-8") as f:
    lector = csv.DictReader(f)
    for fila in lector:
        print(f"{fila['Nombre']} tiene {fila['Edad']} años y estudia {fila['Curso']}")

# Escritura con DictWriter
with open("alumnos_dict.csv", "w", newline='', encoding="utf-8") as f:
    campos = ["Nombre", "Edad", "Curso"]
    escritor = csv.DictWriter(f, fieldnames=campos)
    escritor.writeheader()
    escritor.writerow({"Nombre": "Elena", "Edad": 24, "Curso": "Go"})
    escritor.writerow({"Nombre": "Carlos", "Edad": 25, "Curso": "Rust"})
```

## Lectura de CSV con `pandas`
La librería `pandas` permite leer archivos CSV directamente en un **DataFrame**, una estructura tabular muy potente para análisis de datos.

```python
import pandas as pd

df = pd.read_csv("alumnos.csv")
print("Primeras filas del DataFrame:")
display(df.head())

print("\nInformación general:")
df.info()

print("\nEstadísticas básicas:")
display(df.describe())
```

## Lectura y escritura de archivos JSON
El módulo `json` convierte entre estructuras de Python (`dict`, `list`) y texto JSON (serialización y deserialización).

```python
import json

# Datos de ejemplo
datos = {
    "curso": "Python",
    "estudiantes": [
        {"nombre": "Ana", "edad": 20},
        {"nombre": "Luis", "edad": 22}
    ]
}

# Escribir JSON
with open("datos.json", "w", encoding="utf-8") as f:
    json.dump(datos, f, indent=4, ensure_ascii=False)

# Leer JSON
with open("datos.json", "r", encoding="utf-8") as f:
    contenido = json.load(f)

print("Contenido del JSON:")
print(contenido)
```

## Manejo de errores al trabajar con archivos

Cuando intentas abrir un archivo que no existe o está en una ruta incorrecta, Python lanza una excepción de tipo `FileNotFoundError`.

Para evitar que el programa se detenga bruscamente, podemos usar un bloque `try/except` que capture el error y nos permita manejarlo de forma controlada.

```python
import csv

try:
    with open("archivo_inexistente.csv", "r", encoding="utf-8") as f:
        lector = csv.reader(f)
        for fila in lector:
            print(fila)
except FileNotFoundError:
    print("Error: el archivo 'archivo_inexistente.csv' no se encontró. Verifica la ruta o el nombre.")
```

Ejemplo de código con manejo de múltiples excepciones:

```python
import csv

try:
    with open("datos.csv", "r", encoding="utf-8") as f:
        lector = csv.reader(f)
        for fila in lector:
            print(fila)

except FileNotFoundError:
    print("El archivo no existe. Asegúrate de crearlo antes de leerlo.")

except PermissionError:
    print("No tienes permisos para acceder a este archivo.")

except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
```

## Buenas prácticas
- Usa `with open()` para cerrar archivos automáticamente.
- Especifica la codificación `utf-8`.
- Controla errores con `try/except`.
- Verifica la existencia de archivos antes de escribir o eliminar.
